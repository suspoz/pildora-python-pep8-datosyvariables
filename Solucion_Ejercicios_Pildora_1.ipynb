```python
# Solución Ejercicio 1
total_productos = int("12")
precio_unitario = float("24,95".replace(",", "."))
categoria_producto = "A"
esta_activo = True
descuento_aplicado = bool(0)
peso_kg = float("1.5")

print(total_productos, type(total_productos))
print(precio_unitario, type(precio_unitario))
print(categoria_producto, type(categoria_producto))
print(esta_activo, type(esta_activo))
print(descuento_aplicado, type(descuento_aplicado))
print(peso_kg, type(peso_kg))
```

```python
# Solución Ejercicio 2
# 1. Variables y tipos de datos
unidades_compradas = 6
precio_unidad = 18.75
seccion = "B"
en_stock = True

# 2. Comprobación con type()
print(type(unidades_compradas))
print(type(precio_unidad))
print(type(seccion))
print(type(en_stock))

# 3. Casting de variables
unidades_texto = str(unidades_compradas)
precio_entero = int(precio_unidad)
stock_numero = int(en_stock)

# 4. Casting de valores literales
cantidad_num = int("15")
tarifa_float = float("4.25")
sin_existencias = bool(0)

# 5. Salida de resultados
print(unidades_texto, type(unidades_texto))
print(precio_entero, type(precio_entero))
print(stock_numero, type(stock_numero))
print(cantidad_num, type(cantidad_num))
print(tarifa_float, type(tarifa_float))
print(sin_existencias, type(sin_existencias))
```
