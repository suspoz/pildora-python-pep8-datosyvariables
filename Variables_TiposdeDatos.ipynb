```python
# Variables y tipos de datos (en Python, un carácter es un str de longitud 1)
edad = 25
precio = 9.99
inicial = "A"
activo = True
```

```python
# Comprobar tipos con type()
print(type(edad))     # <class 'int'>
print(type(precio))   # <class 'float'>
print(type(inicial))  # <class 'str'>
print(type(activo))   # <class 'bool'>
```

```python
# Casting (conversión explícita de tipos)
edad_como_texto = str(edad)
precio_entero = int(precio)
texto_a_entero = int("42")
texto_a_flotante = float("3.5")
numero_a_booleano = bool(1)
booleano_a_entero = int(activo)
```

```python
# Visualizamos el contenido final
print(type(edad_como_texto), edad_como_texto)     # <class 'str'> 25
print(type(precio_entero), precio_entero)         # <class 'int'> 9
print(type(texto_a_entero), texto_a_entero)       # <class 'int'> 42
print(type(texto_a_flotante), texto_a_flotante)   # <class 'float'> 3.5
print(type(numero_a_booleano), numero_a_booleano) # <class 'bool'> True
print(type(booleano_a_entero), booleano_a_entero) # <class 'int'> 1
```
